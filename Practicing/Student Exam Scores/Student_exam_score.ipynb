1 . Set up and loading data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# setting visual style for seaborn
sns.set_theme(style="whitegrid")

# loading the generated CSV
df = pd.read_csv("students_score.csv")

# preview first 5 rows and structure
print("------- Data Info ------- ")
df.info()
df.head()
```

2. Pandas practice (Data manipulation and exploration)

```python
# Summary statistics for numerical columns
print("=== Descriptive Statistics === ")
print(df.describe())

# add a calculated column : total score and average score
df["Total_Score"] = df["Math_Score"] + df["English_Score"]
df["Average_Score"] = df["Total_Score"] / 2

# filtering: high performing students (Math >= 80 and english >= 80 )
top_students = df[(df["Math_Score"] >= 80)& (df["English_Score"] >= 80)]
print(f"\nNumber of top-performing students: {len(top_students)}")

# groupby : average math and english scores by grade level
grade_summary = df.groupby("Grade_Level")[["Math_Score", "English_Score"]].mean()
print("\n=== Average Scores by Grade Level ===")
print(grade_summary)

# Value counts: pass/Fail counts by gender
print("\n=== Pass/Fail Breakdown by Gender ===")
print(pd.crosstab(df["Gender"], df["Passed"]))
```
