data cleaning and converting 'data joined' to datetime and extract year

```python
# converting date joined to datetime format
companies["Date Joined"] = pd.to_datetime(companies["Date Joined"])
# creating new column 'year joined' by extracting the year
companies['Year Joined'] = companies["Date Joined"].dt.year
# creating a new column 'month joined' by extracting the month
companies["Month Joined"] = companies['Date Joined'].dt.month
companies.info()
companies.head(5)
```

Sampling and feature engineering 


```python
#Taking a random sample (adjust 'n' depending on dataset size: using min size for safety)
sample_size = min(50,len(companies))
companies_sample = companies.sample(n=sample_size,random_state = 42)
# calculates years taken to reach unicorn status
companies_sample["years_till_unicorn"] = companies_sample["Year Joined"] - companies_sample['Year Founded']
# group by industry and get maximum years taken to become a unicorn
grouped_years = (
  companies_sample[["Industry", "years_till_unicorn", "Valuation","Year Joined"]]
  .groupby("Industry")
  .max()
  .sort_values(by="years_till_unicorn",ascending=False)
)
grouped_years

```

Visualization 1 — Maximum Time to Reach Unicorn Status per Industry

```python
# ploting maximum years to reach unicorn statues per industry
plt.figure(figsize=(10,5))
plt.bar(grouped_years.index, grouped_years["years_till_unicorn"], color="skyblue")

# adding titles and labels
plt.title("Bar plot of maximum years taken by company to become unicorn per industry (from sample)")
plt.xlabel("industry")
plt.ylabel("Maximum number of years")

# Rotate x-axis labels for better readability
plt.xticks(rotation=20, horizontalalignment="right")

# Display plot
plt.show()
```

Clean Valuation Column & Plot Maximum Valuation per Industry

```python
# clean valution column : strip '$' and 'B', convert to int
companies_sample["valuation_billions"] = (
  companies_sample["Valuation"]
  .str.replace("$", "", regex=False)
  .str.replace("B", "", regex=False)
  .astype(int)
)
# group by industry and get the maximum valuation
grouped_val = (
  companies_sample[["Industry","valuation_billions"]]
  .groupby("Industry")
  .max()
  .sort_values(by="valuation_billions")
)
# Plotting maximum valuation per industry
plt.figure(figsize=(10, 5))
plt.bar(grouped_val.index, grouped_val["valuation_billions"])

# Adding titles and labels
plt.title("Bar plot of maximum unicorn company valuation per industry (from sample)")
plt.xlabel("Industry")
plt.ylabel("Maximum valuation in billions of dollars")

# Rotate x-axis labels
plt.xticks(rotation=45, horizontalalignment="right")

# Display plot
plt.show()
```
